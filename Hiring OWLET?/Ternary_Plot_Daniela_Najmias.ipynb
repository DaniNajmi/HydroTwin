```python
import pandas as pd
import plotly.express as px
from google.colab import files
import io
```

```python
# read the excel file
pd.read_excel("Managerial Energy Log.xlsx")
```

```python
df = pd.read_excel("Managerial Energy Log.xlsx")
```

```python
# Look for specific headers and calculate avg
g_avg = df['G (Governance)'].mean()
a_avg = df['A (Audit)'].mean()
c_avg = df['C (Context)'].mean()
```

```python
# Display the results:
print(f"Managerial Audit Summary: ")
print(f"Governance (G) Average: {g_avg}")
print(f"Audit (A) Average: {a_avg}")
print(f"Context (C) Average: {c_avg}")
```

```python
# Generate Ternary Plot
# Put averages into a temporary dataframe for the plot
plot_data = pd.DataFrame({
    'Governance': [g_avg],
    'Audit': [a_avg],
    'Context': [c_avg],
    'Manager': ['Daniela Najmias Lang']
})
```

```python
# plot it out
fig = px.scatter_ternary(
    plot_data,
    a="Governance",
    b="Audit",
    c="Context",
    hover_name="Manager",
    title="Project Managerial Balance Map",
    size=[15], # Makes the dot visible
    color_discrete_sequence=['red'] # can be any color!
)

fig.show()
```
